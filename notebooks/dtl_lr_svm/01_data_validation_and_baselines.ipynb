## Validation strategy

- quick screening: stratified 80/20 holdout;
- model comparison: Stratified 5-Fold CV;
- primary metric: macro F1;
- preprocessing selalu fit hanya pada training fold;
- secondary metrics: F1 per kelas, precision/recall kelas 1, confusion matrix;
- final model diperiksa kembali dengan beberapa random seed.

## Why macro F1?

Kelas positif hanya sekitar 22.22% pada training set. Accuracy dapat terlihat tinggi hanya dengan mengutamakan kelas mayoritas; macro F1 memberi bobot yang sama pada F1 kelas 0 dan kelas 1.
